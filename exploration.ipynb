## Untrained random-policy episode delay

```python
import numpy as np
env = CorridorEnv(demand_profile, seed=1)
obs = env.reset(start_hour=17.5)
rng = np.random.default_rng(0)
for _ in range(200):
    actions = rng.integers(0,2,size=12).tolist()
    obs, r, done, info = env.step(actions)
print(info)
```
